```python
import pandas as pd

# Load the datasets
airports = pd.read_csv(
    "../data/raw/airports.csv",
    low_memory=False
)

qatar = pd.read_csv(
    "../data/processed/qatar_airways_network_seed.csv"
)

print("✈️ AEROFLOW — DATA AUDIT")
print("=" * 60)

print("\nAIRPORT DATASET")
print("-" * 60)
print(f"Rows: {len(airports):,}")
print(f"Columns: {len(airports.columns)}")

print("\nQATAR AIRWAYS NETWORK")
print("-" * 60)
print(f"Rows: {len(qatar):,}")
print(f"Columns: {len(qatar.columns)}")

print("\nQatar Airways columns:")
for column in qatar.columns:
    print(f"  • {column}")
```

```python
# ============================================================
# 1. MISSING VALUE AUDIT
# ============================================================

print("✈️ MISSING VALUE AUDIT")
print("=" * 60)

missing_qatar = qatar.isna().sum()
missing_qatar = missing_qatar[missing_qatar > 0].sort_values(ascending=False)

if len(missing_qatar) == 0:
    print("✅ No missing values in the Qatar Airways dataset.")
else:
    print(missing_qatar)
```

```python
# ============================================================
# 2. DUPLICATE AUDIT
# ============================================================

print("🔍 DUPLICATE AUDIT")
print("=" * 60)

print(f"Duplicate rows: {qatar.duplicated().sum()}")
print(
    f"Duplicate destination airports: "
    f"{qatar['destination_iata'].duplicated().sum()}"
)

print("\nDuplicate destination codes:")
duplicates = qatar[
    qatar["destination_iata"].duplicated(keep=False)
].sort_values("destination_iata")

print(
    duplicates[
        ["origin_iata", "destination_iata", "destination_city"]
    ]
)
```

```python
# ============================================================
# 3. GEOGRAPHIC VALIDATION
# ============================================================

print("🌍 GEOGRAPHIC VALIDATION")
print("=" * 60)

checks = {
    "Missing origin latitude": qatar["origin_latitude"].isna().sum(),
    "Missing origin longitude": qatar["origin_longitude"].isna().sum(),
    "Missing destination latitude": qatar["destination_latitude"].isna().sum(),
    "Missing destination longitude": qatar["destination_longitude"].isna().sum(),
    "Invalid latitude": (
        ((qatar["destination_latitude"] < -90) |
         (qatar["destination_latitude"] > 90)).sum()
    ),
    "Invalid longitude": (
        ((qatar["destination_longitude"] < -180) |
         (qatar["destination_longitude"] > 180)).sum()
    )
}

for check, value in checks.items():
    status = "✅" if value == 0 else "⚠️"
    print(f"{status} {check}: {value}")
```

```python
# ============================================================
# 4. NETWORK DISTRIBUTION
# ============================================================

print("🌍 QATAR AIRWAYS NETWORK DISTRIBUTION")
print("=" * 60)

print("\nDestinations by region:")
print(
    qatar["region"]
    .value_counts()
    .to_string()
)

print("\n\nDestinations by route type:")
print(
    qatar["route_type"]
    .value_counts()
    .to_string()
)
```

```python
# ============================================================
# 5. LONGEST ROUTES FROM DOHA
# ============================================================

print("✈️ LONGEST QATAR AIRWAYS ROUTES FROM DOHA")
print("=" * 60)

longest = (
    qatar[
        [
            "destination_iata",
            "destination_city",
            "destination_country",
            "region",
            "distance_km",
            "route_type"
        ]
    ]
    .sort_values("distance_km", ascending=False)
    .head(15)
)

longest
```

```python
# ============================================================
# 6. FIX QATAR AIRWAYS REGIONS
# ============================================================

# Country-based regional classification
americas = {
    "US", "CA", "MX",
    "BR", "CO", "AR", "CL", "PE", "EC", "VE",
    "BO", "UY", "PY", "CR", "PA", "GT", "DO",
    "JM", "BS", "CU", "TT"
}

europe = {
    "GB", "FR", "DE", "ES", "IT", "NL", "BE",
    "CH", "AT", "IE", "PT", "DK", "SE", "NO",
    "FI", "PL", "CZ", "HU", "RO", "GR", "HR",
    "RS", "BG", "UA", "IS", "LU", "SK", "SI",
    "EE", "LV", "LT", "AL", "BA", "ME", "MK"
}

africa = {
    "ZA", "EG", "MA", "DZ", "TN", "NG", "GH",
    "CI", "SN", "KE", "ET", "UG", "TZ", "RW",
    "ZM", "ZW", "MZ", "AO", "CD", "CG", "CM",
    "BJ", "GN", "SL", "LR", "GM", "CV", "SC",
    "MU", "MG", "NA", "BW"
}

middle_east = {
    "QA", "AE", "SA", "OM", "BH", "KW",
    "IQ", "JO", "LB", "IR", "YE", "IL"
}

caucasus_central_asia = {
    "AM", "AZ", "GE", "KZ", "UZ", "TM", "TJ", "KG"
}

asia_pacific = {
    "IN", "PK", "BD", "LK", "NP", "MV", "BT",
    "CN", "HK", "JP", "KR", "TW", "TH", "VN",
    "MY", "SG", "ID", "PH", "KH", "LA",
    "AU", "NZ", "FJ", "PG"
}


def assign_qatar_region(country):
    if country in americas:
        return "Americas"
    elif country in europe:
        return "Europe"
    elif country in africa:
        return "Africa"
    elif country in middle_east:
        return "Middle East"
    elif country in caucasus_central_asia:
        return "Caucasus & Central Asia"
    elif country in asia_pacific:
        return "Asia-Pacific"
    else:
        return "Other"


qatar["region"] = qatar["destination_country"].apply(
    assign_qatar_region
)

print("✅ Regional classification updated.")
```

```python
# ============================================================
# 7. VERIFY REGIONS
# ============================================================

print("🌍 CORRECTED QATAR AIRWAYS NETWORK")
print("=" * 60)

print(
    qatar[
        [
            "destination_iata",
            "destination_city",
            "destination_country",
            "region"
        ]
    ]
    .sort_values("destination_iata")
    .to_string(index=False)
)
```

```python
print("\n📊 DESTINATIONS BY REGION")
print("=" * 60)

region_counts = (
    qatar["region"]
    .value_counts()
    .rename_axis("Region")
    .reset_index(name="Destinations")
)

region_counts
```

```python
# ============================================================
# 7. REGIONAL DISTRIBUTION
# ============================================================

region_counts = (
    qatar["region"]
    .value_counts()
    .rename_axis("Region")
    .reset_index(name="Destinations")
)

region_counts
```

```python
qatar[
    qatar["region"] == "Americas"
][
    [
        "destination_iata",
        "destination_city",
        "destination_country",
        "distance_km"
    ]
].sort_values("distance_km", ascending=False)
```

```python
# ============================================================
# 8. FINAL REGIONAL DISTRIBUTION
# ============================================================

region_counts = (
    qatar["region"]
    .value_counts()
    .rename_axis("Region")
    .reset_index(name="Destinations")
)

region_counts["Share_%"] = (
    region_counts["Destinations"]
    / region_counts["Destinations"].sum()
    * 100
).round(1)

region_counts
```

```python
# ============================================================
# 9. INVESTIGATE UNCLASSIFIED DESTINATIONS
# ============================================================

other_destinations = qatar[
    qatar["region"] == "Other"
][
    [
        "destination_iata",
        "destination_city",
        "destination_country"
    ]
].sort_values("destination_country")

other_destinations
```

```python
# ============================================================
# 10. FINAL REGION CORRECTIONS
# ============================================================

# Qatar Airways' official network map places these destinations
# in Europe:
europe_corrections = {
    "CY": "Europe",   # Cyprus
    "RU": "Europe",   # Moscow / Russia
    "TR": "Europe"    # Türkiye
}

# Sudan is classified under Africa.
africa_corrections = {
    "SD": "Africa"
}

# Apply corrections
qatar.loc[
    qatar["destination_country"].isin(europe_corrections.keys()),
    "region"
] = qatar["destination_country"].map(europe_corrections)

qatar.loc[
    qatar["destination_country"].isin(africa_corrections.keys()),
    "region"
] = qatar["destination_country"].map(africa_corrections)

print("✅ Final regional corrections applied.")
```

```python
# ============================================================
# 11. FINAL REGION VALIDATION
# ============================================================

print("🌍 FINAL REGION VALIDATION")
print("=" * 60)

other = qatar[qatar["region"] == "Other"]

if len(other) == 0:
    print("✅ No destinations remain in 'Other'.")
else:
    print("⚠️ Destinations still classified as Other:")
    print(
        other[
            [
                "destination_iata",
                "destination_city",
                "destination_country"
            ]
        ]
    )
```

```python
# ============================================================
# 12. FINAL NETWORK REGIONAL DISTRIBUTION
# ============================================================

region_counts = (
    qatar["region"]
    .value_counts()
    .rename_axis("Region")
    .reset_index(name="Destinations")
)

region_counts["Share_%"] = (
    region_counts["Destinations"]
    / region_counts["Destinations"].sum()
    * 100
).round(1)

region_counts
```
